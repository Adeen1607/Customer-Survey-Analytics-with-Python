## Research question

Which departments and products receive the most binary selections in the supplied tables?

## Data and evidence

## Method

Indicator columns are checked as binary and summed. Rankings count selections, not sales units. Original and expanded results must be interpreted separately.

## Interpretation limits

Expanded observations are simulated for visualization practice. Independent random binary values do not reproduce customer correlations. Multiple selections mean shares of selections are not shares of unique customers.

### Guide to code step 1

```python
import pandas as pd
import matplotlib.pyplot as plt
import squarify
```

### Guide to code step 2

**Load the source table.** A DataFrame is a spreadsheet-like table in memory. The path must point to the documented input file; loading a file does not validate its contents.

```python
# Load your Excel files
department_data = pd.read_excel('department_results_expanded.xlsx')
product_data = pd.read_excel('product_results_expanded.xlsx')
```

### Guide to code step 3

**Select values and shape inputs.** `iloc` selects by column position. Check the source schema because a changed column order can select the wrong variable. `reshape(-1, 1)` makes a one-column table expected by a model.

**Summarize observations.** Grouping collects rows with the same label. Counts and sums produce totals; a mean of a binary 0/1 indicator gives a rate. Always identify the denominator and keep missing categories visible.

```python
# Sum of visits for each department and purchases for each product
department_totals = department_data.iloc[:, 2:].sum()
product_totals = product_data.iloc[:, 2:].sum()
```

### Guide to code step 4

This cell prepares named objects used by later steps. Read the assignments from right to left: the expression is evaluated, then stored under the name on the left. Run cells in order so those names exist.

```python
def add_value_labels(ax):
    for p in ax.patches:
        ax.annotate(f'{int(p.get_height())}',
                    (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='center',
                    xytext=(0, 10), textcoords='offset points', fontsize=10, color='black')
```

### Guide to code step 5

**Read the chart.** The axis labels identify units and the grouping defines what is compared. A visual association does not establish cause. Rate charts need denominators; truncated axes can exaggerate small differences.

```python
# Histogram of department visit counts
plt.figure(figsize=(12, 7))
ax1 = department_totals.plot(kind='bar', color='skyblue')
plt.xlabel('Departments')
plt.ylabel('Visit Counts')
plt.title('Distribution of Department Visit Counts')
plt.xticks(rotation=45)
plt.grid(axis='y')
add_value_labels(ax1)
plt.show()
```

### Guide to code step 6

**Select values and shape inputs.** `iloc` selects by column position. Check the source schema because a changed column order can select the wrong variable. `reshape(-1, 1)` makes a one-column table expected by a model.

**Read the chart.** The axis labels identify units and the grouping defines what is compared. A visual association does not establish cause. Rate charts need denominators; truncated axes can exaggerate small differences.

```python
# Treemap of product purchase counts with labels
plt.figure(figsize=(12, 7))
squarify.plot(sizes=product_totals.values,
              label=[f'{label}\n{value}' for label, value in zip(product_totals.index, product_totals.values)],
              alpha=0.8, color=plt.cm.Paired.colors, text_kwargs={'fontsize':10, 'weight':'bold', 'color':'black'})
plt.title('Treemap of Product Purchase Counts')
plt.axis('off')
plt.show()
```

### Guide to code step 7

**Read the chart.** The axis labels identify units and the grouping defines what is compared. A visual association does not establish cause. Rate charts need denominators; truncated axes can exaggerate small differences.

```python
# Top 5 most visited departments
top_5_most_visited_depts = department_totals.nlargest(5)
plt.figure(figsize=(10, 6))
ax3 = top_5_most_visited_depts.plot(kind='bar', color='dodgerblue')
plt.xlabel('Departments')
plt.ylabel('Visit Counts')
plt.title('Top 5 Most Visited Departments')
plt.xticks(rotation=45)
plt.grid(axis='y')
add_value_labels(ax3)
plt.show()
```

### Guide to code step 8

**Read the chart.** The axis labels identify units and the grouping defines what is compared. A visual association does not establish cause. Rate charts need denominators; truncated axes can exaggerate small differences.

```python
# Top 5 most purchased products
top_5_most_purchased_products = product_totals.nlargest(5)
plt.figure(figsize=(10, 6))
ax4 = top_5_most_purchased_products.plot(kind='bar', color='seagreen')
plt.xlabel('Products')
plt.ylabel('Purchase Counts')
plt.title('Top 5 Most Purchased Products')
plt.xticks(rotation=45)
plt.grid(axis='y')
add_value_labels(ax4)
plt.show()
```

### Guide to code step 9

**Read the chart.** The axis labels identify units and the grouping defines what is compared. A visual association does not establish cause. Rate charts need denominators; truncated axes can exaggerate small differences.
